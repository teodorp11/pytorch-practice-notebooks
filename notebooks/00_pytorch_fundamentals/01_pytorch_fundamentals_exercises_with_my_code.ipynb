### 2. Create a random tensor with shape `(7, 7)`.


````python
# Import torch
import torch

# Create random tensor
random_tensor_a = torch.randn(size=(7, 7))

print("random_tensor_a:")
print(random_tensor_a)

print(f"random_tensor_a.shape: {random_tensor_a.shape}\n")

random_tensor_b = torch.arange(start=1, end=50, step=1).reshape(7, 7)

print("random_tensor_b:")
print(random_tensor_b)

print(f"random_tensor_b.shape: {random_tensor_b.shape}\n")
````

### 3. Perform a matrix multiplication on the tensor from 2 with another random tensor with shape `(1, 7)` (hint: you may have to transpose the second tensor).

````python
# Create another random tensor
another_random_tensor = torch.rand(size=(1, 7))

print("another_random_tensor:")
print(another_random_tensor)

print(f"another_random_tensor.shape: {another_random_tensor.shape}\n")

# Perform matrix multiplication
resulted_matrix = torch.matmul(random_tensor_a, another_random_tensor.T)

print("resulted_matrix:")
print(resulted_matrix)

print(f"resulted_matrix.shape: {resulted_matrix.shape}\n")
````
