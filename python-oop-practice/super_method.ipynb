```python
# Link to article: https://realpython.com/python-super/#an-overview-of-pythons-super-function

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * self.length + 2 * self.width
    
    def details(self):
        return f"Rectangle Details\nLength: {self.length}, Width: {self.width}, Perimeter: {self.perimeter()}, Area: {self.area()}\n"

# Here we declare that the Square class inherits from the Rectangle class
class Square(Rectangle):
    def __init__(self, length):
        super().__init__(length, length)
    
    def details(self):
        return f"Square Details\nLength: {self.length}, Perimeter: {super().perimeter()}, Area: {super().area()}\n"

class Cube(Square):
    def surface_area(self):
        face_area = super().area()
        return 6 * face_area
    
    def volume(self):
        face_area = super().area()
        return self.length * face_area

    def details(self):
        return f"Cube Details\nLength: {self.length}, Surface Area: {self.surface_area()}, Volume: {self.volume()}\n"

rectangle_object = Rectangle(3, 4)
print(rectangle_object.details())

square_object = Square(3)
print(square_object.details())

cube_object = Cube(4)
print(cube_object.details())
```
